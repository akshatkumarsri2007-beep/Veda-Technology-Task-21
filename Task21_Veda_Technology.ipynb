```python
import pandas as pd
import sqlite3

sheets = pd.read_excel("/content/northwind_data.xlsx", sheet_name=None)

for name, df in sheets.items():
    print(name, df.shape)

conn = sqlite3.connect(":memory:")
for name, df in sheets.items():
    df.to_sql(name, conn, index=False, if_exists="replace")

def run(query):
    return pd.read_sql_query(query, conn)
```

```python
sheets["customers"].head()
```

```python
run("SELECT * FROM customers")
```

```python
run("SELECT company_name, country FROM customers")
```

```python
run("SELECT * FROM customers WHERE country = 'Germany'")
```

```python
run("SELECT product_name, unit_price FROM products WHERE unit_price > 50")
```

```python
run("SELECT product_name, unit_price FROM products ORDER BY unit_price DESC")
```

```python
run("SELECT company_name, city FROM customers ORDER BY company_name ASC")
```

```python
run("SELECT order_id, order_date FROM orders WHERE strftime('%Y', order_date) = '1997'")
```

```python
run("SELECT product_name, unit_price FROM products ORDER BY unit_price DESC LIMIT 5")
```

```python
run("SELECT company_name, country FROM customers WHERE country IN ('USA', 'UK')")
```

```python
run("SELECT first_name, last_name, city FROM employees WHERE city = 'London'")
```
